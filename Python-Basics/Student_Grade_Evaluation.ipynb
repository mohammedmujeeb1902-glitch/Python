Student Grade Evaluation System

This project demonstrates how to build a simple grade evaluation system in Python using user input, exception handling, conditional statements, and the match-case statement.

The program accepts a student's score, validates the input, assigns a grade, and provides appropriate feedback based on the performance.

```python
def student_score():

    # Exception Handling
    # Accept user input and convert it to an integer
    # If the user enters a non-numeric value, ValueError is raised

    try:
        score = int(input("Please enter your score:"))

    # Handles invalid input such as letters or special characters

    except ValueError:
        print('Invalid input. Please enter a number between 0 and 100')
        return

    # Input Validation
    # Ensure the score is within the valid range (0-100)

    if score < 0 or score > 100:
        print("Invalid score. Please enter a number between 0 and 100")
        return

    # Conditional Statements (if-elif-else)
    # Assign grades based on score ranges

    if 90 <= score <= 100:
        grade = 'A'
    elif 80 <= score <= 89:
        grade = 'B'
    elif 70 <= score <= 79:
        grade = 'C'
    elif 60 <= score <= 69:
        grade = 'D'
    else:
        grade = 'F'

    print(f"Your grade is { grade} ")

    # Match-Case Statement
    # Execute different actions based on the grade value
    
    match grade:
        case "A":
            print("Excellent")
        case "B" | "C":
            print("Good")
        case "D" | "F":
            print("You need improvement")

def main():
    student_score()
       
if __name__ == '__main__':
    main()
```
