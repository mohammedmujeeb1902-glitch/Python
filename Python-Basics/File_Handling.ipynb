 # Student Information File Handling System

<!-- Objective
This program demonstrates:

Collecting student names and scores using user input
Validating names and numeric scores
Writing student records to a text file (students.txt)
Reading data back from the file
Displaying student records with serial numbers using enumerate() -->

<!-- Concepts Used
Functions
Loops (for, while)
Input validation
String methods (strip(), isalpha(), capitalize())
File handling (open(), read(), write())
Conditional statements (if-else)
Enumeration (enumerate()) -->

<!-- Workflow
Enter details for 3 students.
Validate the student name.
Validate the student score.
Save the data to students.txt.
Read the file content.
Display the records in a numbered format. -->

```python
def students_info():
    with open("students.txt", "w") as file:
        for i in range(3):
            
            while True:
                name = input("Enter student name: ").strip()
                if name.isalpha():
                    name = name.capitalize()
                    break
                else:
                    print("Please enter a valid name .")

            
            while True:
                score = input(f"Enter score for {name}: ").strip()
                if score.isdigit():
                    score = int(score)
                    break
                else:
                    print("Please enter a valid numeric score.")

            
            file.write(f"{name} - {score}\n")


def execute_info():
    with open("students.txt", "r") as file:
        for idx, line in enumerate(file, start=1):
            print(f"{idx}. {line.strip()}")


def main():
    students_info()
    print("\nReading back from file:\n")
    execute_info()


if __name__ == "__main__":
    main()
```
