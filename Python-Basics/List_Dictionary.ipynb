# Student Performance Analysis Using Comprehensions

# Objective

<!-- This project demonstrates the use of List Comprehensions and Dictionary Comprehensions in Python to analyze student performance data efficiently. -->

```python
# The below variable 'students' is list of dictionaries

students = [ 
    {"name": "Drake", "score": 98}, 
    {"name": "Milner", "score": 69}, 
    {"name": "Harvey", "score": 97}, 
    {"name": "Roger", "score": 70},
    {"name": "Han", "score": 35},
    ]

def main():

    # using for loop and getting student score more than 80 marks in a List Comprehension

    toppers = [student["name"] for student in students if student["score"] > 90]
    print("Toppers are: ", toppers)
    
    # using for loop and getting student score more than or equal to 40 marks in a List Comprehension

    results = {student["name"]: ("Pass" if student["score"] >= 40 else "Fail") for student in students}
    print("Results are: ", results)

if __name__ == '__main__':
    main()
```
